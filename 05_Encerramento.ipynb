## 1. Apresentação final

Os alunos devem apresentar o qur fizeram no projeto da disciplina. Roteiro sugerido para apresntação de 10–15 minutos:

1. necessidade do cliente e área-alvo;
2. requisitos e critérios de aceite;
3. planejamento e execução;
4. pipeline e decisões técnicas;
5. resultados de completude e acurácia;
6. nuvem/modelo e casos de uso $\leftarrow$ aproveitar bem esta parte!
7. limitações e riscos residuais
8. recomendação para trabalhos futuros.

Mostre conceitos, ideias, valores humanos, imagens, não telas cheias de parâmetros. 
Se colocar métricas, prefira gráficos focando na comparação, no antes e depois, nos ganhos.
Se colocar as limitações, faça com comentários objetivos e sem arrodeios.


## 2. Demo 3D e vídeo flythrough

Prepare uma cena leve, câmera inicial, pontos de interesse, legenda/escala, classes e instruções de navegação. Teste na máquina e na rede da apresentação. Tenha três níveis de contingência: viewer ao vivo, vídeo gravado e imagens estáticas.

O flythrough deve orientar o público: visão geral → zonas críticas → comparação/medição → produto final. Evite movimentos rápidos, cortes sem contexto e compressão que esconda defeitos.

## 3. Pacote de entrega

Estrutura sugerida:

```text
ENTREGA_SLAM100/
├── 01_relatorio/
├── 02_nuvem/
├── 03_modelos_3d/
├── 04_controle_qualidade/
├── 05_visualizacao/
├── 06_metadados/
├── README.md
```

Inclua licença/restrições de uso, CRS e referencial vertical, unidades, convenção de nomes, softwares testados, versões, limites conhecidos e contato técnico. Dados brutos só entram se o contrato e a política de segurança permitirem.

## 4. Aceite e transferência

O **termo de aceite** deve relacionar cada requisito ao produto, evidência e resultado. Registre versões entregues, meio/data da transferência, responsável, pendências, prazo de retenção e aprovação do cliente.

## 5. Lições aprendidas

Conduza uma retrospectiva sem buscar culpados:

- o que foi planejado e o que ocorreu?
- quais condições favoreceram ou degradaram o SLAM?
- onde houve retrabalho e qual foi sua causa?
- quais parâmetros, trajetos ou verificações devem virar padrão?
- que risco não foi previsto?
- qual evidência faltou para decidir mais cedo?
