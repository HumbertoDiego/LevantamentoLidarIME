Inclua licença/restrições de uso, CRS e referencial vertical, unidades, convenção de nomes, softwares testados, versões, limites conhecidos e contato técnico. Dados brutos só entram se o contrato e a política de segurança permitirem.

````python
from pathlib import Path
import hashlib
import pandas as pd

PASTA_ENTREGA = Path('produtos')  # ajuste sem apontar para os dados brutos

def sha256_arquivo(caminho, bloco=1024 * 1024):
    digest = hashlib.sha256()
    with caminho.open('rb') as arquivo:
        for parte in iter(lambda: arquivo.read(bloco), b''):
            digest.update(parte)
    return digest.hexdigest()

def criar_manifesto(pasta):
    if not pasta.exists():
        return pd.DataFrame(columns=['arquivo', 'bytes', 'sha256'])
    arquivos = sorted(p for p in pasta.rglob('*') if p.is_file())
    return pd.DataFrame([
        {'arquivo': p.relative_to(pasta).as_posix(), 'bytes': p.stat().st_size, 'sha256': sha256_arquivo(p)}
        for p in arquivos
    ])

manifesto = criar_manifesto(PASTA_ENTREGA)
manifesto.head(20)
````

A célula anterior apenas calcula o manifesto em memória. Depois de revisar a pasta, salve explicitamente com `manifesto.to_csv('produtos/MANIFESTO_SHA256.csv', index=False)`. Gere o manifesto por último ou exclua o próprio manifesto do cálculo para evitar autorreferência.
